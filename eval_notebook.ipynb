```python
import pandas as pd
import ast

%pwd
```

```python
test_df = pd.read_csv("experiments/baseline/results/Llama-3.2-1B/df.csv")
```

```python
test_df['tasks'] = test_df['tasks'].apply(ast.literal_eval).apply(lambda x: x.pop(0) if len(x) > 0 else None)
test_df['malicious_labels'] = test_df['malicious_labels'].apply(ast.literal_eval).apply(lambda x: x.pop(0) if len(x) > 0 else None)
```

```python
accuracy_per_task = test_df.groupby('tasks').apply(
    lambda x: pd.Series({
        'accuracy': (x['predicted_labels'] == x['malicious_labels']).mean(),
        'count': len(x)
    })
).reset_index()
```

```python
sum(test_df.malicious_labels == test_df.predicted_labels)/len(test_df.predicted_labels) * 100
```
